## 1. 실습 방법 <small>(실습)</small>

CodeMate에 주는 지시문은 다섯 칸으로 씁니다. 무엇을 원하는지, 어떤 조건인지, 어떤 모양으로 받을지, 헷갈릴 점과 하지 말 것을 미리 적어 두면 원하는 코드를 받기 쉽습니다.

```text
[역할·목표] 무엇을 하는 사람으로서, 무엇을 만들어 달라는지
[배경·제약] 어느 셀의 어느 빈칸인지, 쓸 수 있는 것과 없는 것
[출력 형식] 코드만? 코드 + 설명 3줄?
[모호함 줄이기] 헷갈릴 만한 부분을 미리 정해 주기
[하지 말 것] 빈칸 밖은 고치지 않기, 새 기능 추가하지 않기
```

순서는 늘 같습니다.
1. 빈칸 위의 TODO 주석을 읽고, 무엇을 채워야 하는지 한 문장으로 말해 봅니다.
2. 빈칸이 든 코드 셀을 선택한 상태에서 지시문을 CodeMate 채팅에 붙여 넣습니다.
3. CodeMate가 채운 코드를 읽고, 모르는 줄이 있으면 "이 줄이 하는 일"을 다시 묻습니다.
4. 실행해 봅니다. 오류가 나면 오류 메시지를 그대로 CodeMate에 붙여 넣고 원인부터 설명해 달라고 합니다.
5. 셀 아래 "정답 보기"를 펼쳐 비교합니다. 다르다고 틀린 것은 아닙니다. 왜 다른지 말할 수 있으면 됩니다.

## 6. 05_hitl_interactive <small>(코드) (실습)</small>

(작성 예정)

준비 코드입니다. 필요한 도구를 불러오고 모델을 준비합니다. 접혀 있으니 펼치지 않아도 되고, 실행만 합니다.

````python
from typing import Annotated

from dotenv import load_dotenv
from langchain_core.messages import HumanMessage, ToolMessage
from langchain_core.tools import tool
from langgraph.checkpoint.memory import MemorySaver
from langgraph.graph import START, StateGraph
from langgraph.graph.message import add_messages
from langgraph.prebuilt import ToolNode, tools_condition
from typing_extensions import TypedDict

from config.llm import get_llm

load_dotenv()

llm = get_llm()


@tool
def send_email(recipient: str, subject: str, body: str) -> str:
    """이메일을 전송한다."""
    print(f'\n  [실제 전송] {recipient} / 제목: {subject}')
    return f'이메일이 {recipient}에게 성공적으로 전송되었습니다.'


tools = [send_email]
llm_with_tools = llm.bind_tools(tools)
tool_node = ToolNode(tools)


class State(TypedDict):
    messages: Annotated[list, add_messages]


def agent_node(state: State) -> dict:
    return {'messages': [llm_with_tools.invoke(state['messages'])]}


builder = StateGraph(State)
builder.add_node('agent', agent_node)
builder.add_node('tools', tool_node)
builder.add_edge(START, 'agent')
builder.add_conditional_edges('agent', tools_condition)
builder.add_edge('tools', 'agent')

app = builder.compile(
    checkpointer=MemorySaver(),
    interrupt_before=['tools'],
)


def print_tool_calls(tool_calls: list[dict]) -> None:
    print('\n실행 대기 중인 도구 호출')
    for index, tool_call in enumerate(tool_calls, start=1):
        print(f'{index}. {tool_call["name"]}')
        for key, value in tool_call['args'].items():
            print(f'   - {key}: {value}')


def reject_tool_calls(config: dict, tool_calls: list[dict], reason: str) -> None:
    rejection_messages = [
        ToolMessage(
            content=f'사용자가 도구 실행을 거부했습니다. 사유: {reason}',
            tool_call_id=tool_call['id'],
        )
        for tool_call in tool_calls
    ]
    app.update_state(config, {'messages': rejection_messages}, as_node='tools')


def ask_human_decision() -> tuple[str, str]:
    print('\n도구를 실행할까요?')
    print('- y: 승인 후 실행')
    print('- n: 거부 후 에이전트에게 취소 사유 전달')
    print('- q: 프로그램 종료')

    decision = input('선택(y/n/q): ').strip().lower()
    if decision == 'n':
        reason = input('거부 사유를 입력하세요: ').strip()
        return decision, reason
    return decision, ''


def run_agent(user_input: str, thread_id: str = 'hitl_demo') -> None:
    config = {'configurable': {'thread_id': thread_id}}
    result = app.invoke({'messages': [HumanMessage(content=user_input)]}, config=config)

    while True:
        state = app.get_state(config)
        tool_calls = state.values['messages'][-1].tool_calls

        if not tool_calls:
            print('\n에이전트 응답')
            print(result['messages'][-1].content)
            break

        print_tool_calls(tool_calls)
        decision, reason = ask_human_decision()

        if decision == 'y':
            result = app.invoke(None, config=config)
        elif decision == 'n':
            reject_tool_calls(config, tool_calls, reason or '사유 없음')
            result = app.invoke(None, config=config)
        elif decision == 'q':
            print('프로그램을 종료합니다.')
            break
        else:
            print('y, n, q 중 하나를 입력하세요.')


if __name__ == '__main__':
    request = input('에이전트에게 요청할 내용을 입력하세요: ')
    run_agent(request)
````
