## A2A의 배경과 MCP와의 관계 · PDF 1–5쪽

[원본 1~5쪽](./3-3-a2a-protocol-understanding.pdf#page=1)

서로 다른 역할·환경의 Agent가 협업하려면 상대 내부 구현을 몰라도 능력을 찾고 작업을 위임할 수 있어야 합니다. MCP는 Agent와 도구·데이터 연결, A2A는 Agent 간 작업 위임입니다. A2A로 요청받은 Agent가 내부에서는 MCP Tool을 쓸 수도 있습니다. 4쪽의 배달 앱·배달 관리·가게 주문 관리 Agent 도식과 5쪽의 주문·재고·POS 서비스 도식에서 Agent 사이 연결과 Agent-도구 연결을 나누어 읽어 보세요.


## 협업 데이터의 단위 · PDF 6–8쪽

[원본 6~8쪽](./3-3-a2a-protocol-understanding.pdf#page=6)

**Agent Card**는 이름, 설명, 제공 Skill, 입출력 방식, 연결 정보를 알립니다. **Task**는 위임된 일의 단위, **Message**는 요청과 중간 대화, **Artifact**는 보고서·코드 같은 산출물입니다. 긴 작업은 상태와 메시지 이력을 함께 추적합니다.


## 서버: 실행, 소개, 라우팅 · PDF 9–12쪽

[원본 9~12쪽](./3-3-a2a-protocol-understanding.pdf#page=9)

슬라이드에서 `AgentExecutor`는 요청을 내부 Agent 실행으로 연결하고 결과 이벤트를 냅니다. `AgentCard`는 `/.well-known/agent-card.json`에서 발견됩니다. `RequestHandler`는 실행기·작업 저장소·큐와 라우트를 묶습니다. 교육용 구조 설명이므로 현재 설치 SDK와 클래스 이름이 일치하는지 실습 전에 확인하세요.


## 클라이언트: 발견 후 요청 · PDF 13–15쪽

[원본 13~15쪽](./3-3-a2a-protocol-understanding.pdf#page=13)

클라이언트는 `A2ACardResolver`로 Card를 읽고 연결 준비를 합니다. 그 뒤 `SendMessageRequest`에 사용자 메시지를 넣어 보내며, 응답 스트림의 Message에서 결과를 읽습니다. 슬라이드는 SDK 1.x 헤더와 JSON-RPC 전송 예시를 보여줍니다.


### 실행해 보기: Agent Card 검사

외부 서버 없이 Card의 필수 정보를 살펴보는 작은 예제입니다.


```python
card = {
    "name": "Researcher Agent",
    "description": "주제를 조사해 요약합니다.",
    "skills": [{"id": "web_research", "name": "웹 리서치"}],
    "url": "http://localhost:8000/",
}
required = {"name", "description", "skills", "url"}
missing = required - card.keys()
assert not missing, f"빠진 항목: {missing}"
print(card["name"], "제공 기능:", [s["name"] for s in card["skills"]])
```
