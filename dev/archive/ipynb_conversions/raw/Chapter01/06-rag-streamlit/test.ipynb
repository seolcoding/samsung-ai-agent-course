## 원본 코드 — 수동 실행

아래 셀은 `.py` 전체를 그대로 보존한다. 위에서 아래로 한 번에 실행되므로 환경과 부작용을 확인한다. **자동 실행/Run All을 사용하지 않는다.**

**이 코드셀의 실행 조건·부작용:** 버튼 클릭과 클릭 횟수 상태를 다루는 Streamlit UI 예제다. 원본 .py를 Streamlit으로 실행해 확인한다.

**특별 주의:** 실습본은 session_state로 횟수를 보존하고 정답본은 일반 변수를 사용해 재실행 때 초기화된다.



```python
import streamlit as st

st.title("버튼 누르면 스크립트 전체 다시 실행")

if "clicks" not in st.session_state:  
    st.session_state.clicks = 0

if st.button(' 클릭 '):
    st.session_state.clicks +=1
    st.balloons()

st.write(f"풍선 발사 {st.session_state.clicks}회")
```
