````python
def chatbot_node(state: State) -> dict:
    response = llm.invoke(state["messages"])
    return {"messages": [response]}


builder = StateGraph(State)
builder.add_node("chatbot", chatbot_node)
builder.add_edge(START, "chatbot")
builder.add_edge("chatbot", END)
app = None  # 아래에서 보기를 고르기 전 상태
````

`detect_language` 노드는 LLM에게 언어를 물은 뒤 답(`result.content`)을 `"ko"`나 `"en"`으로 정리해 `lang`에 담습니다. LLM은 `ko`만 답하지 않고 `Ko.`나 `언어: ko`처럼 답하기도 합니다.

**✍️ CHOOSE_HERE: 어떤 답이 와도 `lang`을 `"ko"`나 `"en"`으로 정리하는 줄은?**

| 보기 | 코드 |
|---|---|
| A | `lang = result.content` |
| B | `lang = "ko" if "ko" in result.content.lower() else "en"` |
| C | `lang = "ko" if result.content == "ko" else "en"` |
| D | `lang = "en" if "en" in result.content else "ko"` |

````python
# ✍️ CHOOSE_HERE: 먼저 ? 자리에 고른 보기를 적고 실행합니다
MY_CHOICE = "?"

compare_options(
    {
        "A": 'lang = result.content',
        "B": 'lang = "ko" if "ko" in result.content.lower() else "en"',
        "C": 'lang = "ko" if result.content == "ko" else "en"',
        "D": 'lang = "en" if "en" in result.content else "ko"',
    },
    samples=[("ko", "ko"), ("Ko.", "ko"), ("언어: ko", "ko"), ("en", "en"), ("EN", "en")],
    my_choice=MY_CHOICE,
)
````

소문자로 바꾼 뒤 `ko`가 들어 있는지만 봅니다. 답에 다른 글자가 섞여도 둘 중 하나로 정리됩니다. A는 답을 그대로 담아 `Ko.` 같은 값이 라우팅 함수로 갑니다. C는 정확히 `ko`일 때만 맞습니다. D는 대문자 `EN`을 놓칩니다.

````python
@tool
def get_exchange_rate(base_currency: str, target_currency: str) -> float:
    """고정 환율을 반환한다. base_currency와 target_currency는 USD, KRW, EUR 중 하나."""
    rates = {("USD", "KRW"): 1350.0, ("KRW", "USD"): 1 / 1350.0, ("EUR", "KRW"): 1470.0,
             ("KRW", "EUR"): 1 / 1470.0, ("USD", "EUR"): 0.92, ("EUR", "USD"): 1.09}
    if base_currency == target_currency:
        return 1.0
    return rates.get((base_currency, target_currency), -1.0)


@tool
def calculate(expression: str) -> float:
    """간단한 수식을 계산한다. 예: '1350 * 100'"""
    allowed = set("0123456789+-*/(). ")
    if not all(c in allowed for c in expression):
        raise ValueError("허용되지 않는 문자가 포함되어 있습니다.")
    return eval(expression)  # noqa: S307


tools = [get_exchange_rate, calculate]
llm_with_tools = get_llm().bind_tools(tools)


def agent_node(state: State) -> dict:
    return {"messages": [llm_with_tools.invoke(state["messages"])]}
````

**✍️ CHOOSE_HERE: ToolNode로 tools 노드를 추가하고, 도구 실행 뒤 agent로 돌아가게 잇는 묶음을 고르세요.**

줄 끝 번호가 같은 세 줄이 한 보기입니다. 세 줄을 함께 드래그해 선택하고 `Ctrl+/`를 누릅니다. `tools_condition`은 도구 호출이 있으면 `"tools"`라는 이름의 노드로 보냅니다.

- ⛔ `[ToolMessage]`에서 끝나고 최종 답이 없다 → 도구 실행 뒤 agent로 돌아가는지 확인합니다
- ⛔ `unknown target 'tools'` 오류가 난다 → 노드 이름을 확인합니다

````python
builder = StateGraph(State)
builder.add_node("agent", agent_node)
builder.add_edge(START, "agent")
builder.add_conditional_edges("agent", tools_condition)  # → 'tools' or END

# ✍️ CHOOSE_HERE: 같은 번호(1️⃣~4️⃣)가 붙은 세 줄을 함께 골라 주석을 풉니다
# tool_node = ToolNode(tools)               # 1️⃣
# builder.add_node("tools", tool_node)      # 1️⃣
# builder.add_edge("agent", "tools")        # 1️⃣
#
# tool_node = ToolNode(tools)               # 2️⃣
# builder.add_node("tools", tool_node)      # 2️⃣
# builder.add_edge("tools", "agent")        # 2️⃣
#
# tool_node = ToolNode(tools)               # 3️⃣
# builder.add_node("tool_runner", tool_node)# 3️⃣
# builder.add_edge("tool_runner", "agent")  # 3️⃣
#
# tool_node = ToolNode(tools)               # 4️⃣
# builder.add_node("tools", tool_node)      # 4️⃣
# builder.add_edge("tools", END)            # 4️⃣

app = builder.compile()
result = app.invoke({"messages": [HumanMessage(content="100달러를 원화로 환산하면 얼마야?")]})
for msg in result["messages"]:
    print(f"  [{msg.__class__.__name__}] {msg.content or msg.tool_calls}")
````
