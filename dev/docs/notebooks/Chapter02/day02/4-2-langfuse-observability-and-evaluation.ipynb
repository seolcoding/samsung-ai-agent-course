## 관측이 필요한 이유 · PDF 1–5쪽

[원본 1~5쪽](../../../raw/Chapter02/day02/4-2-langfuse-observability-and-evaluation.pdf#page=1)

LLM 애플리케이션의 한 번의 요청 안에도 프롬프트·모델·도구 호출이 여러 번 들어갈 수 있습니다. Langfuse는 실행 흐름, 지연시간, 비용, 오류를 기록해 어느 단계에서 문제가 났는지 살피게 합니다. 슬라이드는 클라우드와 자체 설치 방식도 소개합니다.


## Trace, CallbackHandler, @observe · PDF 6–8쪽

[원본 6~8쪽](../../../raw/Chapter02/day02/4-2-langfuse-observability-and-evaluation.pdf#page=6)

Trace는 하나의 요청에서 벌어진 단계를 재구성합니다. `CallbackHandler`는 LangChain·LangGraph 이벤트를 연결하고 `config={"callbacks": [...]}`로 전달합니다. `@observe()`는 프레임워크 바깥의 일반 Python 함수도 같은 관측 흐름에 넣습니다. 입력·출력 기록 범위에는 민감 정보가 들어가지 않도록 살펴야 합니다.


## Score와 규칙 기반 평가 · PDF 9–11쪽

[원본 9~11쪽](../../../raw/Chapter02/day02/4-2-langfuse-observability-and-evaluation.pdf#page=9)

Trace가 **무슨 일이 있었나**를 기록한다면 Score는 **결과가 기준을 충족했나**를 남깁니다. 규칙 기반 평가는 출처 표기·키워드·길이 같은 조건에 빠르고 재현 가능하게 답합니다. 문장의 유용성이나 근거 충실성 전체를 대신하지는 못합니다.


## LLM-as-a-Judge · PDF 12–13쪽

[원본 12~13쪽](../../../raw/Chapter02/day02/4-2-langfuse-observability-and-evaluation.pdf#page=12)

별도 LLM에게 관련성·근거 충실성·완성도 같은 기준을 평가하게 할 수 있습니다. 점수 범위와 채점 기준을 고정하고, 항목별 근거를 요구하며, 가능하면 규칙 기반 결과와 함께 검토합니다. 평가 모델의 출력도 재검토 대상입니다.


### 실행해 보기: 간단한 규칙 점수

같은 답변에 대해 어떤 기준을 적용했는지 명시합니다.


```python
def has_source(answer: str) -> float:
    return 1.0 if "출처" in answer or "http" in answer else 0.0

answers = ["결론입니다.", "결론입니다. 출처: 내부 문서 A"]
for answer in answers:
    print({"answer": answer, "source_score": has_source(answer)})
```
