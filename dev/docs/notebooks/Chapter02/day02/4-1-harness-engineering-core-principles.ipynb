## 왜 실행 구조를 설계하는가 · PDF 1–6쪽

[원본 1~6쪽](../../../raw/Chapter02/day02/4-1-harness-engineering-core-principles.pdf#page=1)

긴 코드 작성·리서치는 중간 상태, 도구 선택, 검증, 실패 대응이 필요합니다. 슬라이드의 구분은 Prompt=질문 표현, Context=참고 정보, Harness=목표를 수행하고 검증하도록 만드는 운영 구조입니다. 모델의 능력만으로 작업 완료를 보장할 수 없습니다.


## 하네스의 제어 구조 · PDF 7–10쪽

[원본 7~10쪽](../../../raw/Chapter02/day02/4-1-harness-engineering-core-principles.pdf#page=7)

하네스는 모델·도구·메모리·역할·흐름·검증을 함께 설계합니다. 기본 순서는 **목표 정의 → 작업 분해 → 도구 연결 → 검증 설계 → 예외 처리**입니다. 완료 조건, 허용 도구, 멈춤·재시도·사람 확인 조건을 먼저 적으면 위임 범위가 분명해집니다.


## 단순하게 시작하고 구조를 고르기 · PDF 11–15쪽

[원본 11~15쪽](../../../raw/Chapter02/day02/4-1-harness-engineering-core-principles.pdf#page=11)

슬라이드는 단순한 프롬프트·검색·단일 도구로 먼저 해결 가능한지 살피라고 합니다. Workflow는 경로가 미리 정해져 재현성과 테스트가 중요할 때, Agent는 중간 관찰에 따라 다음 행동을 바꿔야 할 때 맞습니다. 복잡한 구조는 문제 해결 근거가 있을 때 추가합니다.


## SKILL.md의 역할과 구성 · PDF 16–20쪽

[원본 16~20쪽](../../../raw/Chapter02/day02/4-1-harness-engineering-core-principles.pdf#page=16)

Skill은 특정 작업에서 Agent가 읽는 절차입니다. Description, When to use, Constraints와 Workflow, Examples, Verification을 구분해 적습니다. 적용 조건은 구체적으로, 절차는 실제 행동 순서로, 검증은 관찰 가능한 결과로 써야 합니다. 판단 불가 또는 실패 시 행동도 넣습니다.
