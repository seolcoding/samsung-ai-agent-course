## 1. AI, 머신러닝, LLM (슬라이드 2–14)

**AI**는 지능적 행동을 다루는 큰 분야이고, **머신러닝**은 데이터에서 규칙을 학습해 새 입력을 예측하는 방법입니다. 슬라이드의 아이스크림 예에서는 기온과 판매량의 관계를 데이터에서 찾습니다. 데이터 수집 → 모델 선택 → 파라미터 조정 → 새 기온 예측 순서입니다.

**LLM**도 학습된 모델입니다. 말뭉치에서 문맥 다음에 올 토큰의 패턴을 배우고, 지시 형식의 데이터로 추가 조정되어 질문에 답합니다. 그럴듯한 문장과 사실 검증은 다른 일입니다. 계산·최신 정보·출처가 중요한 답은 따로 확인해야 합니다. 슬라이드의 “앵무새” 비유는 다음 말 예측을 기억하기 위한 비유입니다.

### 작은 실험: 데이터로 선형 관계 추정 (슬라이드 5–8)

아래 코드는 기온과 판매량의 관계를 최소제곱 직선으로 근사합니다. 실제 판매를 설명하는 충분한 모델이라는 뜻은 아닙니다. 입력이 바뀌면 예측도 바뀌는 모습을 살펴보세요.

```python
temperatures = [20, 22, 24, 26, 28]
sales = [40, 47, 54, 61, 68]
x_bar = sum(temperatures) / len(temperatures)
y_bar = sum(sales) / len(sales)
slope = sum((x-x_bar)*(y-y_bar) for x,y in zip(temperatures,sales)) / sum((x-x_bar)**2 for x in temperatures)
intercept = y_bar - slope*x_bar
prediction = slope*30 + intercept
print(f"학습된 관계: 판매량 ≈ {slope:.1f} × 기온 + {intercept:.1f}")
print(f"30도 예측 판매량: {prediction:.1f}")
```

**생각하기:** 데이터가 20–28도뿐인데 40도를 예측한다면 어떤 위험이 있을까요? “학습 데이터 안에서 맞았다”와 “새 상황에서도 맞는다”를 구별해 설명해 보세요.

## 2. 문장 생성에서 함수 호출로 (슬라이드 15–19)

LLM의 응답은 기본적으로 텍스트입니다. 도구 호출에서도 모델은 **도구 이름과 인자**를 구조화해 제안하고, 애플리케이션이 이를 검사해 함수를 실행합니다. 슬라이드의 `get_weather` 예처럼 사용할 수 있는 도구와 입력 형식을 문맥에 제공해야 합니다. 실행 결과는 다시 모델 입력으로 전달할 수 있습니다.

```python
def get_weather(location: str) -> str:
    samples = {"Seoul": "맑음", "Dongtan": "흐림"}
    return samples.get(location, "샘플 데이터 없음")

request = {"tool": "get_weather", "arguments": {"location": "Seoul"}}  # 모델이 제안했다고 가정
allowed_tools = {"get_weather": get_weather}
name = request["tool"]
if name not in allowed_tools:
    raise ValueError(f"허용되지 않은 도구: {name}")
result = allowed_tools[name](**request["arguments"])
print("도구 관찰 결과:", result)
```

**검토:** 위 코드에서 모델이 실제로 함수를 호출했나요? `arguments`에 잘못된 키가 들어오면 무슨 일이 생기나요? 도구 실행 전에 이름·인자·권한을 검증하는 이유를 말해 보세요.

## 3. Agent의 반복 흐름 (슬라이드 20–22)

Agent는 목표를 위해 판단(Reason) → 도구 선택·실행(Action) → 결과 확인(Observation)을 반복할 수 있습니다. 종료 조건이 없으면 쓸데없는 반복이 생기므로 최대 횟수와 성공 기준을 정해야 합니다. **LangChain**은 프롬프트·모델·도구 등의 연결, **LangGraph**는 상태를 가진 분기·반복 흐름의 구성에 쓰입니다.

```python
observations = []
for step in range(1, 4):
    action = "검색" if not observations else "답변"
    observation = "근거 문서 1건 발견" if action == "검색" else "근거를 포함한 답변 작성"
    observations.append(observation)
    print(f"{step}회: 판단 → {action} → {observation}")
    if action == "답변":
        break
```
