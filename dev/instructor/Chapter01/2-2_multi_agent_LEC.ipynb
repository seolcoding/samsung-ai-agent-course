## 2. Multi-Agent 패턴 <small>(슬라이드) (퀴즈)</small>

### 2.1 Supervisor <small>(슬라이드) (퀴즈)</small>

```python
show_slides("2-2", [7, 8])
```

Supervisor 패턴은 중앙 조정 Agent가 작업을 분석하고, 알맞은 하위 Agent에게 일을 나눠 주는 구조입니다.
Supervisor는 전체 목표와 진행 상태를 관리합니다. 하위 Agent는 조사, 작성, 검토, 계산처럼 한 가지 역할에 집중합니다.
최종 응답은 Supervisor가 하위 Agent의 결과를 종합해 작성합니다.
다만 모든 결정이 중앙 조정자를 거치므로 조정자가 병목이 될 수 있습니다. 복잡한 작업일수록 설계가 중요합니다.

**왜 배우나:** Multi-Agent 흐름을 읽을 때 가장 먼저 볼 것은 "다음에 누가 일할지를 누가 정하는가"입니다. Supervisor 패턴에서는 그 결정을 한 Agent가 맡습니다.

```python
show_quiz("../assets/quizzes/2-2/q21.json")
```

**정리:** Supervisor가 목표와 진행 상태를 쥐고 하위 Agent에 일을 나눕니다. 조정자가 병목이 되지 않게 설계해야 합니다.

<details><summary>강사용 메모</summary>

- 슬라이드는 "최종 응답은 Supervisor가 종합해 작성"이라고 합니다. 실습의 `02_supervisor.py`에서는 Writer가 초안(`draft`)을 쓰고, Supervisor는 다음 차례(researcher/writer/FINISH)만 정합니다. 최종 출력은 Writer의 초안입니다. 수강생이 물으면 이 차이를 짚어 줍니다.
- 실습 코드는 Supervisor의 결정을 구조화 출력(`RouteDecision`, `with_structured_output`)으로 받아 선택지를 세 값으로 묶습니다. 슬라이드에는 없는 내용입니다(커리큘럼 2.3 "Structured Output(RouteDecision)으로 라우팅 결정 안정화").

</details>

### 2.2 Hierarchical <small>(슬라이드)</small>

```python
show_slides("2-2", [9])
```

Hierarchical 패턴은 Agent 조직을 계층으로 나눕니다.
상위 Agent는 큰 목표와 정책을 관리하고, 중간 Agent는 하위 작업을 다시 나눠 담당자를 정하며, 하위 Agent는 구체적인 실행과 결과 생성을 맡습니다.
Supervisor 패턴에서 중앙 조정자 한 명이 지던 부담을 중간 Agent들이 나눠 가지는 구조입니다.

**정리:** 상위(목표·정책) → 중간(분해·배정) → 하위(실행)로 층을 나눠 조정 부담을 나눕니다.

### 2.3 Network <small>(슬라이드) (퀴즈)</small>

```python
show_slides("2-2", [10])
```

Network 패턴은 중앙 관리자 없이 Agent들이 서로 직접 메시지를 주고받는 구조입니다.
각 Agent는 필요한 상대에게 직접 요청하거나 결과를 넘깁니다. 창의적 토론, 상호 검토, 분산 협업에 쓰입니다.
중앙 병목은 줄지만 전체 흐름을 통제하기 어렵습니다. 자율성이 큰 만큼 주고받는 규칙이 중요합니다.

```python
show_quiz("../assets/quizzes/2-2/q23.json")
```

**정리:** Network는 병목이 적은 대신 흐름을 통제하기 어렵습니다. 그래서 규칙이 중요합니다.

## 3. 협업 구조 설계 <small>(슬라이드) (퀴즈)</small>

### 3.1 역할 분리와 책임 경계 <small>(슬라이드) (퀴즈)</small>

```python
show_slides("2-2", [11, 12])
```

Multi-Agent 설계의 핵심은 Agent를 많이 만드는 것이 아니라 책임을 잘 나누는 것입니다.
업무 목적, 필요한 도구, 판단 기준을 기준으로 나누고, 각 Agent가 처리할 입력과 돌려줄 출력의 형식을 정해 둡니다.
책임 경계가 모호하면 Multi-Agent도 단일 Agent보다 나아지지 않습니다.

좋은 역할 분리는 프롬프트와 평가 기준을 단순하게 만듭니다.
- Researcher는 근거 수집 품질로 평가합니다.
- Writer는 구조와 표현 품질로 평가합니다.
- Reviewer는 누락, 오류, 근거 일치 여부로 평가합니다.

**왜 배우나:** 역할이 분명하면 결과를 검토하는 기준도 분명해집니다. AI가 만든 결과물을 리뷰할 때도 "이 역할이 맡은 일을 했는가"로 나눠 보면 무엇을 고칠지 빨리 찾을 수 있습니다.

```python
show_quiz("../assets/quizzes/2-2/q31.json")
```

**정리:** 목적·도구·판단 기준으로 책임을 나누고, 입력과 출력 형식을 정하면 역할마다 평가 기준이 단순해집니다.

<details><summary>강사용 메모</summary>

- 실습의 선택 장(`03_supervisor_with_critic.py`)에 나오는 Critic이 슬라이드의 Reviewer 역할입니다. 사실 오류, 논리 모순, 핵심 요구 누락만 보고 문체는 무시하도록 프롬프트에 적혀 있습니다.

</details>

### 3.2 단일 Agent와 Multi-Agent 비교 <small>(슬라이드) (퀴즈)</small>

```python
show_slides("2-2", [13])
```

Multi-Agent가 항상 단일 Agent보다 좋은 선택은 아닙니다.
단순 질의응답이나 짧은 작업은 단일 Agent가 더 빠르고 저렴합니다. 역할이 뚜렷하게 나뉘고 검토가 필요하면 Multi-Agent가 유리합니다.

구조는 문제의 복잡도와 운영 요구사항을 보고 고릅니다.
- 작업 단계가 적고 도구가 제한적이면 단일 Agent를 먼저 고려합니다.
- 조사와 작성처럼 결과물이 다른 단계가 있으면 역할 분리를 고려합니다.
- 위험한 결정이나 품질 검증이 필요하면 Reviewer Agent를 추가합니다.

Agent가 늘면 비용과 지연 시간도 늘어납니다. 그만큼 품질이 좋아졌는지 측정해야 합니다.

**왜 배우나:** 4일차 캡스톤의 기본 구조는 Supervisor가 Researcher와 Writer에게 일을 나누는 멀티 에이전트 시스템입니다. 어떤 역할을 따로 떼고 어디에 검토를 넣을지 정할 때 이 기준을 씁니다.

```python
show_quiz("../assets/quizzes/2-2/q32.json")
```

**정리:** 짧고 단순한 일은 단일 Agent, 결과물이 다른 단계가 있으면 역할 분리, 검증이 필요하면 Reviewer를 더합니다. 늘어난 비용만큼 효과를 측정합니다.

<details><summary>강사용 메모</summary>

- 캡스톤 연결 근거: 커리큘럼 4.3 "스켈레톤 코드 통합 아키텍처: Supervisor → Researcher(MCP+RAG+웹검색) → HITL 승인 → Writer". 캡스톤은 자유 주제와 스켈레톤 중 고를 수 있습니다.
