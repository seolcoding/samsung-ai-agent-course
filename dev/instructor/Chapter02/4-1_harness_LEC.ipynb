### 3.2 Workflow와 Agent <small>(슬라이드) (퀴즈)</small>

```python
show_slides("4-1", [13, 14, 15])
```

**Workflow**는 실행 경로가 미리 정해진 구조입니다. 개발자가 단계, 분기 조건, 도구 호출 순서를 미리 설계하고, LLM은 각 단계에서 분류·요약·추출·생성처럼 제한된 역할을 합니다. 예측하기 쉽고 테스트하기 쉽습니다.

**Agent**는 실행 경로를 모델이 상황에 따라 정하는 구조입니다. 목표는 주어지지만 세부 단계는 실행 중에 정해집니다. Agent는 관찰 결과를 보고 다음 행동을 고르고, 도구 사용·정보 수집·계획 수정을 되풀이합니다.

| | Workflow를 고를 때 | Agent를 고를 때 |
|---|---|---|
| 순서 | 실행 순서가 분명하다 | 다음 행동을 미리 알기 어렵다 |
| 계획 | 분기 기준을 미리 정할 수 있다 | 환경의 반응에 따라 계획을 바꿔야 한다 |
| 중요한 것 | 재현성과 테스트 가능성 | 동적인 도구 선택과 탐색 |
| 알맞은 일 | 문서 요약 후 형식 변환, 입력 유형 분류 후 전용 처리, 코드 생성 후 테스트 실행, 표준 템플릿 기반 산출물 작성 | 원인을 알 수 없는 오류 디버깅, 여러 자료를 탐색하는 리서치, 중간 판단이 필요한 모호한 작업, 새 정보에 따라 계획을 바꾸는 긴 작업 |

**왜 배우나:** 캡스톤 프로젝트의 흐름(Supervisor → Researcher → 사람 승인 → Writer)은 정해진 흐름 안에 스스로 판단하는 Agent를 넣은 구조입니다. 어느 부분을 고정하고 어느 부분을 모델에게 맡길지 고르는 기준이 이 표입니다.

```python
show_slides("4-1", [16, 17])
```

SKILL.md는 Agent가 특정 작업을 할 때 참고하는 절차 문서입니다.

- 이 능력을 **언제** 써야 하는지 알려 줍니다.

Skill은 하네스 안에서 Agent의 행동을 안정시키는 운영 규칙입니다.

**왜 배우나:** 작업별 지침을 문서로 나눠 두면, 코드와 모델을 그대로 두고 문서만 고쳐 Agent의 동작을 바꿀 수 있습니다. 캡스톤 프로젝트에서도 Researcher와 Writer에 역할별 SKILL.md를 붙입니다.

### 4.3 좋은 SKILL.md의 작성 기준과 순서 <small>(슬라이드) (퀴즈)</small>

```python
show_slides("4-1", [19, 20])
```

좋은 Skill은 Agent가 스스로 판단하고 실행할 수 있을 만큼 구체적이어야 합니다.

- 적용 조건을 모호하게 쓰지 않습니다.
- 수행 절차를 실제 행동 순서대로 씁니다.
- 필요한 파일, 도구, 명령, 출력 위치를 분명히 적습니다.
- 성공 조건과 검증 방법을 넣습니다.
- 실패하거나 판단하기 어려울 때 어떻게 할지 적습니다.

작성 순서는 다섯 단계입니다.

1. **목적**: Skill이 해결하려는 작업과 적용 범위를 한 문단으로 정합니다.
2. **조건**: 언제 이 Skill을 쓰고, 언제 쓰지 않는지 나눕니다.
3. **절차**: Agent가 실제로 따를 순서를 단계별로 씁니다.
4. **도구**: 필요한 도구, 파일, 명령, 출력 위치를 구체적으로 적습니다.
5. **검증**: 최종 산출물의 품질 기준과 확인 방법을 적어, 다시 검토할 수 있게 합니다.
