### 1.1 LLM, Agent, LangChain <small>(슬라이드) (퀴즈)</small>

```python
show_slides("3-1", [3, 4])
```

LLM이 어떻게 컴퓨터를 쓰는 Agent가 되었는지를 네 낱말로 정리합니다.

- **LLM**: 방대한 언어 데이터로 "이 다음엔 어떤 말이 나올까?"를 예측하는 모델입니다.
- **Function(Tool) Calling**: LLM의 출력을 해석해 컴퓨터의 기능을 실행하는 일입니다.
- **Agent**: 함수를 호출하고 그 결과를 검토하는 과정을 반복해, 복잡한 작업을 처리하도록 구성한 LLM 파이프라인입니다.
- **LangChain(LangGraph)**: LLM을 활용한 워크플로를 만드는 데 쓰는 라이브러리입니다.

슬라이드 아래 코드는 LangChain의 기본 모양입니다. 프롬프트 템플릿(`prompt`), 모델(`model`), 출력 파서(`parser`)를 `|`로 이어 체인 하나를 만들고, `invoke`에 `{"topic": "LLM"}`을 넣으면 `{topic}` 자리가 채워진 질문이 모델로 가서 답변 텍스트가 나옵니다.

**왜 배우나:** 과정 2의 MCP와 A2A는 모두 "도구를 호출하고 결과를 검토하는 반복" 위에 올라갑니다. Agent의 정의를 먼저 맞춰 두면 새 주제가 어디에 붙는지 보입니다.

```python
show_quiz("../assets/quizzes/3-1/q11.json")
```

**정리:** LLM은 다음 말을 예측하고, Tool Calling은 그 출력으로 기능을 실행하며, Agent는 호출과 검토를 반복합니다. LangChain은 이 흐름을 만드는 라이브러리입니다.

### 1.2 RAG <small>(슬라이드) (퀴즈)</small>

```python
show_slides("3-1", [5])
```

RAG는 LLM 응답의 품질을 올리는 기법 중 하나입니다. 세 단계로 이뤄집니다.

1. 문서를 쪼개고, 의미가 얼마나 비슷한지 비교할 수 있는 형태로 바꿉니다.
2. 사용자 질문과 가장 관련이 높은 문서를 검색합니다.
3. LLM이 답변을 만들 때 검색한 문서 내용을 더해 답변 품질을 높입니다.

슬라이드 화면은 과정 1에서 Streamlit으로 만든 보험 약관 QA 챗봇입니다. 메리츠 펫보험 약관을 근거로 답변합니다.

```python
show_quiz("../assets/quizzes/3-1/q12.json")
```

**정리:** RAG는 문서를 쪼개 검색할 수 있게 바꾸고, 질문과 가까운 문서를 찾아, 답변에 그 내용을 더합니다.

### 1.3 복잡한 Agent 표현을 위한 LangGraph <small>(슬라이드) (퀴즈)</small>

```python
show_slides("3-1", [6])
```

LangGraph는 Agent의 실행 흐름을 그래프로 표현해, 복잡한 분기와 반복을 다룰 수 있게 합니다.

- **State**를 중심으로 실행 맥락을 관리합니다.
- **Node**와 **Edge**로 Agent 흐름을 구성합니다.
- **Conditional Edge**로 조건에 따라 갈 길을 나눕니다.
- **ToolNode**, **Human-in-the-loop** 같은 Agent 패턴을 표현합니다.

슬라이드 그림은 과정 1의 멀티 에이전트 구조입니다. 사용자 요청을 Supervisor가 받아 Research, Writer, Critic Agent에게 일을 나누고, 최종 답변을 만듭니다.

```python
show_quiz("../assets/quizzes/3-1/q13.json")
```

**정리:** LangGraph는 State로 맥락을 관리하고, Node·Edge로 흐름을 짜고, Conditional Edge로 분기합니다.

### 2.2 LangGraph 심화 <small>(슬라이드) (퀴즈)</small>

```python
show_slides("3-1", [9])
```

LangGraph 심화에서는 Agent 실행 흐름을 더 안정적이고 확장하기 쉬운 구조로 다듬습니다.

- **Subgraph**로 그래프를 모듈로 나눕니다.

### 2.3 Harness Engineering <small>(슬라이드) (퀴즈)</small>

```python
show_slides("3-1", [10])
```

Harness Engineering은 Agent가 안정적으로 일할 수 있도록 **실행 환경과 맥락을 설계**하는 관점입니다.

- 필요한 컨텍스트를 어떻게 제공할지 설계합니다.
- 도구를 쓰는 조건과 실패했을 때 복구하는 전략을 정합니다.
- Skill, 지침, 파일 구조를 Agent가 읽기 쉽게 정리합니다.

**왜 배우나:** 과정 2의 실습은 AI와 함께 코드를 만들고 결과를 검토하는 방식으로 진행합니다. AI에게 어떤 맥락과 지침을 주느냐에 따라 결과가 달라지므로, 그 맥락을 설계하는 일이 곧 Harness Engineering의 대상입니다.

```python
show_quiz("../assets/quizzes/3-1/q23.json")
```

**정리:** Harness Engineering은 모델을 바꾸지 않고, 컨텍스트·도구 사용 규칙·지침과 파일 구조를 설계해 Agent를 안정시킵니다.

### 2.4 Observability <small>(슬라이드) (퀴즈)</small>

```python
show_slides("3-1", [11])
```

Observability는 Agent가 실제로 어떻게 실행됐는지 **추적하고 평가**하는 역량입니다.

- **Trace**로 모델 호출, 도구 실행, 체인 흐름을 확인합니다.
- **Score**로 응답 품질을 수치로 나타냅니다.
- **Langfuse**로 실행 로그와 평가 결과를 관리합니다.
- 이렇게 모은 기록으로 RAG와 Agent 응답을 계속 개선합니다.

```python
show_quiz("../assets/quizzes/3-1/q24.json")
```

**정리:** Trace로 실행을 보고, Score로 품질을 재고, Langfuse로 둘을 관리해 응답을 개선합니다.
