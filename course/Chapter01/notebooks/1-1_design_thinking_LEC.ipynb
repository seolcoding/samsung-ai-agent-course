### 4.2 다음 단어 예측기의 단점 <small>(슬라이드) (퀴즈)</small>

```python
show_slides("1-1", [12])
```

LLM에 "To be or not to be,"를 넣으면 "that is the question"을 이어 씁니다. LLM은 "다음 단어 예측기"입니다.
LLM의 여러 단점도 여기서 나옵니다.

### 4.3 Instruction Tuning과 챗봇 <small>(슬라이드)</small>

```python
show_slides("1-1", [13, 14])
```

이렇게 만든 모델을 질문과 답변 형식의 데이터로 한 번 더 조정합니다. 이것을 Instruction Tuning이라고 합니다.
입력은 질문, 출력은 답변이 되도록 학습시키면 우리가 아는 LLM 챗봇(ChatGPT, Gemini, Claude 등)이 됩니다.
챗봇도 여전히 앞서 나온 말과 관련이 깊은 말을 출력합니다. 그래서 입력이 풍부하고 정확할수록 출력의 품질이 좋아집니다. Prompt Engineering, Context Engineering이 여기서 나옵니다.

슬라이드의 결론처럼 LLM은 앵무새입니다. 다만 세상의 거의 모든 말을 들은 앵무새라서, 내가 한 말과 가장 가까운 답을 줍니다.

**왜 배우나:** "입력이 정확할수록 출력이 좋다"는 AI에게 코드를 부탁할 때도 같습니다. 무엇을, 어떤 조건으로 만들지 분명하게 적을수록 AI가 만든 결과가 좋아집니다.

## 5. LLM의 함수 호출 <small>(슬라이드) (퀴즈)</small>

```python
show_slides("1-1", [15, 16])
```

LLM의 입출력은 기본적으로 문자에서 문자입니다. LLM이 직접 나서서 컴퓨터를 조작하지는 못합니다.
그러면 LLM이 우리 대신 작업을 하게 하려면 무엇이 필요할까요? 이 질문에서 Agent 이야기가 시작됩니다.

```python
show_slides("1-1", [17])
```

Fine-Tuning은 이미 학습한 큰 모델에 특수한 목적의 데이터를 조금 더 학습시키는 것입니다. 이것으로 LLM의 동작을 조정할 수 있습니다.
앞에서 본 Instruction Tuning도 Fine-Tuning의 한 종류입니다. 다음 말을 이어 쓰던 모델을, 앞 글을 질문으로 알아듣고 답하는 모델로 바꿨습니다.
같은 방법으로 LLM의 출력을 "함수를 호출하는 형태"로 조정하면 어떨까요?

```python
show_slides("1-1", [18])
```

그것이 Function Calling입니다.

1. 특정 맥락에서 함수를 호출하는 출력을 만들도록 LLM을 Fine-Tuning합니다.
2. 요청을 보낼 때 쓸 수 있는 함수 목록을 맥락에 함께 넣습니다.
3. 사용자의 입력에 함수가 필요해 보이면, LLM은 슬라이드의 `get_weather` 예시처럼 함수 이름과 인자를 적은 텍스트를 출력합니다.
4. LLM 바깥의 프로그램이 이 출력을 읽고 적힌 함수를 대신 호출합니다.

LLM은 함수를 직접 실행하지 않습니다. "어떤 함수를 어떤 인자로 호출할지"를 글로 쓸 뿐이고, 실행은 그 글을 읽은 프로그램이 합니다.

```python
show_quiz("../../assets/quizzes/1-1/q_fc.json")
```

## 6. Agent <small>(슬라이드) (퀴즈)</small>

```python
show_slides("1-1", [20])
```

LLM은 맥락이 많을수록 더 정확한 출력을 만듭니다. 그렇다면 최종 답을 내기 전에 LLM이 맥락을 스스로 만들게 하면 어떨까요?

- **Reason:** 작업을 위해 무슨 행동을 해야 할지 생성합니다.
- **Action:** 그 출력을 바탕으로 호출할 함수를 정합니다.
- **Observation:** 함수의 결과를 다시 입력으로 넣어 다음 출력을 만듭니다.

이 세 단계를 목표를 이룰 때까지 되풀이합니다. 이것이 ReAct 프레임워크입니다.
이처럼 LLM의 출력 흐름을 알맞게 조정해 스스로 목표를 이루도록 만든 AI를 **Agent**라고 합니다.

**왜 배우나:** 뒤에서 배울 LangGraph에서 이 Reason → Action → Observation 반복을 그래프로 직접 만듭니다.

## 7. LangChain과 LangGraph <small>(슬라이드) (퀴즈)</small>

```python
show_slides("1-1", [21, 22])
```

AI Agent를 만든다는 것은 세 가지를 하는 일입니다.

1. LLM에 들어갈 입력을 조정하고
2. LLM의 출력을 바탕으로 현실의 도구를 호출하고
3. 그 결과의 흐름을 조정해 복잡한 문제를 풀게 합니다.

이 일을 돕는 프레임워크가 두 가지입니다.

- **LangChain:** 사용자 프롬프트, LLM, 외부 도구 등을 이어 파이프라인을 만들 수 있게 해 줍니다.
- **LangGraph:** LLM을 쓰는 워크플로우를 그래프 형태로 만들 수 있게 해 줍니다.

**왜 배우나:** 두 프레임워크는 이 과정 실습의 중심 도구입니다. 코드는 AI가 대부분 써 주지만, 코드를 읽을 줄 알아야 하는 이유가 있습니다.

- 설계는 사람이 합니다.
- 코드를 알면 API 오류나 할루시네이션의 원인을 빨리 찾습니다.
- 코드 구조를 알고 지시하면 결과가 좋고, 엉뚱한 재시도로 버리는 토큰이 줄어듭니다.
