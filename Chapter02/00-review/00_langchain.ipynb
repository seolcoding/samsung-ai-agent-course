## 코드 흐름

**읽을 때 확인할 점:** stream의 chunk는 완전한 문장이 아닐 수 있다. 두 호출은 독립적인 요청이다.


## 실행 전 확인

LLM에 두 번 접속한다. 사내 기본 게이트웨이는 사내망에서만 동작한다.

아래 원본 코드 셀은 실행을 직접 선택해야 하는 `manual-run` 셀입니다. 실행 준비 셀은 프로젝트 루트를 찾고 원본 `.py`의 경로를 `__file__`에 설정합니다.


```python
from pathlib import Path
import os
import sys

_repo_root = next(
    (p for p in (Path.cwd(), *Path.cwd().parents)
     if (p / "pyproject.toml").is_file() and (p / "Chapter02").is_dir()),
    None,
)
if _repo_root is None:
    raise FileNotFoundError("저장소 안에서 이 노트북을 열어 주세요.")
os.chdir(_repo_root)
if str(_repo_root) not in sys.path:
    sys.path.insert(0, str(_repo_root))
__file__ = str(_repo_root / "Chapter02/00-review/00_langchain.py")
```

## 원본 코드

다음 셀에는 원본 `.py` 전체가 그대로 들어 있습니다.


```python
from dotenv import load_dotenv
from langchain_core.messages import HumanMessage, SystemMessage

from config.llm import get_llm

load_dotenv()


llm = get_llm()

# ── 기본 invoke ───────────────────────────────────────────────────────────────
# SystemMessage: LLM의 역할/행동 지침 설정
# HumanMessage: 사용자 입력
messages = [
    SystemMessage(content='너는 초보자를 위한 AI 튜터야.'),
    HumanMessage(content='LLM이 무엇인지 3문장으로 설명해줘.'),
]

response = llm.invoke(messages)
print(response.content)
print()

# ── Streaming ─────────────────────────────────────────────────────────────────
# stream()은 토큰 단위로 응답을 yield함
# end='' 로 줄바꿈 없이 실시간 출력
for chunk in llm.stream(messages):
    print(chunk.content, end='', flush=True)
print()
```
