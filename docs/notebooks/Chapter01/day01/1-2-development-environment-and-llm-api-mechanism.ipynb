## 1. 학습과 추론 (슬라이드 3–7)

**사전 학습**은 대규모 텍스트의 패턴을 익히는 단계이고, **Instruction Tuning**은 요청을 수행하는 대화 형식으로 행동을 조정하는 단계입니다. 추론 때 모델은 완성된 문장을 저장소에서 꺼내지 않고, 현재 문맥을 바탕으로 다음 토큰을 하나씩 선택합니다. API는 메시지와 설정을 받고 이 생성 과정을 응답으로 돌려줍니다.

```python
candidates = {"brown": 0.7, "gray": 0.2, "orange": 0.1}
next_token = max(candidates, key=candidates.get)
print("The quick", next_token, "fox")
print("이 딕셔너리는 개념 설명용 숫자이며 실제 LLM 확률이 아닙니다.")
```

## 2. 토큰, 비용, 컨텍스트 윈도우 (슬라이드 8–14)

토큰은 모델이 처리하는 텍스트 단위이며 단어와 정확히 일치하지 않습니다. 시스템 지시, 사용자 질문, 이전 대화, 검색·도구 결과가 모두 입력 컨텍스트에 포함됩니다. 출력도 토큰 단위로 생성됩니다. 긴 입력은 비용과 지연을 늘리고 컨텍스트 한도에 닿을 수 있습니다. 필요한 자료만 선별하거나 이전 대화를 요약하는 이유입니다.

아래의 공백 분리는 **실제 토크나이저가 아닙니다.** “무엇이 컨텍스트에 들어가는가”만 시각화합니다.

```python
messages = [
    {"role": "system", "content": "문서에 없는 정보는 모른다고 답하라."},
    {"role": "user", "content": "Agent의 상태란 무엇인가?"},
    {"role": "assistant", "content": "현재 작업을 이어 가는 데 필요한 데이터입니다."},
]
for message in messages:
    print(f"{message['role']:9} | {len(message['content'])}글자 | {message['content']}")
print("총 글자 수:", sum(len(x["content"]) for x in messages))
```

**검토:** 글자 수와 실제 토큰 수는 왜 다를까요? 도구가 50쪽 문서를 반환했다면 모두 다음 요청에 넣기보다 어떤 부분을 남겨야 할까요?

## 3. 입력 품질과 대화 구조 (슬라이드 15–18)

좋은 입력에는 목표, 필요한 배경, 제약, 출력 형식, 검토 기준이 들어갑니다. Chat API의 `system`은 전반적 규칙, `user`는 요청, `assistant`는 앞선 응답을 나타냅니다. 역할이 붙은 메시지 목록은 대화의 흐름을 보존합니다. 실제 API마다 필드 이름이나 우선순위 규칙이 다를 수 있으므로 연결할 모델의 문서를 확인하세요.

```python
request = [
    {"role": "system", "content": "근거가 부족하면 확인 필요라고 답한다."},
    {"role": "user", "content": "목표: RAG 설명. 대상: 입문자. 형식: 두 문장. 근거: 검색된 문서만 사용."},
]
for item in request:
    assert item["role"] in {"system", "user", "assistant"}
print(request)
```
