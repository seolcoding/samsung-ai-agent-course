## 1. 단일 Agent의 한계 (슬라이드 3–6)

조사·계획·작성·검토를 한 흐름에 모두 넣으면 현재 역할과 실패 원인을 구분하기 어려워집니다. 작업 자료와 도구 설명이 늘면 컨텍스트가 커지고 유사 도구 사이의 선택도 어려워집니다. 역할 분리는 이 문제를 줄일 수 있지만 비용·지연도 늘립니다.

## 2. 협업 패턴 (슬라이드 7–10)

| 패턴 | 누가 다음 일을 정하는가 | 적합한 상황 | 주의점 |
|---|---|---|---|
| Supervisor | 중앙 조정 Agent | 조사·작성·검토의 순서가 뚜렷함 | 중앙 병목 |
| Hierarchical | 상위·중간 Agent가 단계별 배분 | 작업이 여러 수준으로 나뉨 | 계층 설계 비용 |
| Network | Agent끼리 직접 요청 | 상호 검토·토론 | 흐름 추적과 통제 |

패턴은 이름보다 **책임과 메시지 흐름**으로 비교하세요.

## 3. 책임 경계 (슬라이드 11–12)

각 Agent에 입력, 할 일, 반환 형식, 검토 기준을 적습니다. Researcher는 근거 수집, Writer는 설명 구성, Reviewer는 누락·오류 확인에 집중할 수 있습니다. Reviewer가 새 사실을 지어내는 식으로 경계가 흐려지면 역할 분리의 이점이 사라집니다.

```python
roles = {
    "researcher": {"input": "질문", "output": "근거 목록", "check": "출처가 있는가"},
    "writer": {"input": "질문+근거 목록", "output": "답변 초안", "check": "근거를 인용했는가"},
    "reviewer": {"input": "초안+근거 목록", "output": "검토 결과", "check": "근거 밖 주장이 없는가"},
}
for name, contract in roles.items():
    print(name, "→", contract)
```

### 작은 협업 시뮬레이션

아래는 LLM Agent가 아니라 순수 함수로 **인계 계약**을 점검하는 예제입니다. 실제 Agent로 바꿀 때도 출처 목록이 Writer와 Reviewer까지 전달되는지 확인하세요.

```python
def researcher(question):
    return {"question": question, "evidence": ["슬라이드 12: 역할 분리와 책임 경계"]}
def writer(packet):
    return {**packet, "draft": "역할마다 입력과 출력 계약을 정합니다."}
def reviewer(packet):
    ok = bool(packet["evidence"]) and bool(packet["draft"])
    return {**packet, "approved": ok}
result = reviewer(writer(researcher("협업 역할은 어떻게 나누나요?")))
print(result)
```
