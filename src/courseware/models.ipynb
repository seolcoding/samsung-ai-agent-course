## 작은 실행과 검토

입력·출력·실패 조건을 예상하고 실행하세요. 원문 코드의 구현과 실제 결과를 대조합니다.

```python
from langchain.agents import create_agent
from langchain_core.tools import tool
from courseware.models import ScriptedMultiplyModel
@tool
def multiply(a: int, b: int) -> int:
    """정수 두 개를 곱합니다."""
    return a * b
agent = create_agent(model=ScriptedMultiplyModel(), tools=[multiply])
result = agent.invoke({"messages": [("human", "6과 7을 곱하세요")]})
print(result["messages"][-1].content)
assert "42" in result["messages"][-1].content
```
