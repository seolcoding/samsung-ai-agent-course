## 작은 실행과 검토

입력·출력·실패 조건을 예상하고 실행하세요. 원문 코드의 구현과 실제 결과를 대조합니다.

```python
from fastmcp import Client
from courseware.mcp_server import build_mcp
async with Client(build_mcp()) as client:
    print("도구:", [t.name for t in await client.list_tools()])
    result = await client.call_tool("search_knowledge", {"query": "RAG", "k": 1})
    print(result.data)
    assert result.data[0]["id"] == "D1"
```
